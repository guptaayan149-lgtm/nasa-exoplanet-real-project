```python
import lightkurve as lk
import matplotlib.pyplot as plt

search = lk.search_lightcurve("TIC 111464926", mission="TESS")

lc = search.download_all().stitch()

clean_lc = lc.remove_nans()

clean_lc = clean_lc.remove_outliers()

flat_lc = clean_lc.flatten(window_length=101)

flat_lc.plot()
plt.title("HAT-P-7 b - Flattened Light Curve")
plt.show()

period = 2.2047

folded_lc = flat_lc.fold(period=period)

folded_lc.plot()
plt.title("HAT-P-7 b - Phase Folded Light Curve")
plt.show()
```
