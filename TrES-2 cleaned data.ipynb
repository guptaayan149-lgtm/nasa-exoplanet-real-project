```python
import lightkurve as lk
import matplotlib.pyplot as plt

search_result = lk.search_lightcurve("TrES-2", mission="TESS")
lc = search_result.download()
lc.plot()
plt.title("TrES-2 Light Curve")
plt.show()
```

```python
clean_lc = lc.remove_nans()
clean_lc = clean_lc.remove_outliers()

clean_lc.plot()
plt.title("TrEs-2 - CLeaned Light Curve")
plt.show()
```

```python
flat_lc = clean_lc.flatten(window_length=301)

flat_lc.plot()
plt.title("TrES-2 - flattened graph")
plt.show()
```

```python
period = 2.47
folded_lc = flat_lc.fold(period=period)
folded_lc.plot()
plt.title("TrES-2 folded graph")
plt.show()
```
