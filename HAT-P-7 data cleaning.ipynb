```python
import lightkurve as lk
import matplotlib.pyplot as plt

search_result = lk.search_lightcurve("HAT-P-7", mission="TESS")
lc = search_result.download()

lc.plot()
plt.title("HAT-P-7 - Original graph")
plt.show()
```

```python
clean_lc = lc.remove_nans()
clean_lc = clean_lc.remove_outliers()

clean_lc.plot()
plt.title("HAT-P-7 - CLeaned Light Curve")
plt.show()
```

```python
flat_lc = clean_lc.flatten(window_length=301)

flat_lc.plot()
plt.title("HAT-P-7 flattened graph")
plt.show()
```

```python
period = 2.2047
folded_lc = flat_lc.fold(period=period)
folded_lc.plot()
plt.title("HAT-P-7 folded graph")
plt.show()
```
