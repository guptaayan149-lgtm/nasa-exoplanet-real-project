```python
import lightkurve as lk
import matplotlib.pyplot as plt

# Star 1: HAT-P-7
search_result = lk.search_lightcurve("HAT-P-7", mission="TESS")
lc = search_result.download()
lc.plot()
plt.title("HAT-P-7 Light Curve")
plt.show()

# Star 2: WASP-18
search_result = lk.search_lightcurve("WASP-18", mission="TESS")
lc = search_result.download()
lc.plot()
plt.title("WASP-18 Light Curve")
plt.show()

# Star 3: TrES-2
search_result = lk.search_lightcurve("TrES-2", mission="TESS")
lc = search_result.download()
lc.plot()
plt.title("TrES-2 Light Curve")
plt.show()
```
