```python

import lightkurve as lk
import matplotlib.pyplot as plt

search_result = lk.search_lightcurve("Kepler-10", mission="Kepler")
lc_collection = search_result.download_all()

lc = lc_collection.stitch()
clean_lc = lc.remove_nans().remove_outliers()

# 1. Normal light curve (cleaned)
clean_lc.plot()
plt.title("Kepler-10 Normal Light Curve")
plt.show()

# 2. Flattened light curve
flat_lc = clean_lc.flatten(window_length=301)

flat_lc.plot()
plt.title("Kepler-10 Flattened Light Curve")
plt.show()

# 3. Phase-folded light curve
period = 0.837491

folded_lc = flat_lc.fold(period=period)

folded_lc.plot()
plt.title("Kepler-10 b Phase-Folded Light Curve")
plt.show()
```
