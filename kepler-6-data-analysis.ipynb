```python
import lightkurve as lk
import matplotlib.pyplot as plt

search_result = lk.search_lightcurve("Kepler-6", mission="Kepler")
lc_collection = search_result.download_all()

lc = lc_collection.stitch()
clean_lc = lc.remove_nans().remove_outliers()

# Graph 1: Normal light curve
clean_lc.plot()
plt.title("Kepler-6 Normal Light Curve")
plt.show()

# Graph 2: Flattened light curve
flat_lc = clean_lc.flatten(window_length=301)

flat_lc.plot()
plt.title("Kepler-6 Flattened Light Curve")
plt.show()

# Graph 3: Phase-folded light curve
period = 3.234699

folded_lc = flat_lc.fold(period=period)

folded_lc.plot()
plt.title("Kepler-6 b Phase-Folded Light Curve")
plt.show()
```
