```python
import lightkurve as lk
import matplotlib.pyplot as plt

search = lk.search_lightcurve("HAT-P-7", mission="TESS", author="SPOC")

print(search)

lc = search[0].download(download_dir="tess_data")

lc.plot()
plt.title("HAT-P-7 Original Light Curve")
plt.show()

clean_lc = lc.remove_nans()
clean_lc = clean_lc.remove_outliers()
flat_lc = clean_lc.flatten(window_length=101)

flat_lc.plot()
plt.title("HAT-P-7 Cleaned and Flattened Light Curve")
plt.show()
```
