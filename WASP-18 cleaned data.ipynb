```python
import lightkurve as lk
import matplotlib.pyplot as plt

search_result = lk.search_lightcurve("WASP-18", mission="TESS")
lc = search_result.download()

lc.plot()
plt.title("WASP-18 - Original graph")
plt.show()
```

```python
clean_lc = lc.remove_nans()
clean_lc = clean_lc.remove_outliers()

clean_lc.plot()
plt.title("WASP-18 - CLeaned Light Curve")
plt.show()
```

```python
flat_lc = clean_lc.flatten(window_length=101)

flat_lc.plot()
plt.title("WASP-18 flattened graph")
plt.show()
```

```python
period = 0.94
folded_lc = flat_lc.fold(period=period)
folded_lc.plot()
plt.title("WASP-18 folded graph")
plt.show()
```
